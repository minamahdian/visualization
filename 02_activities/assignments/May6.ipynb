```python
import pandas as pd
df=pd.read_csv('https://raw.githubusercontent.com/prasertcbs/basic-dataset/master/movie_quotes.csv')

# Assuming 'line' is the column containing text for the wordcloud
# Instead of 'line', use the actual column name containing the movie lines.
# Based on the DataFrame preview, it seems the column is named 'quote'.
text=" ".join(df['quote'].astype(str).tolist())  # Join all lines from the 'quote' column into a single string

wordcloud = WordCloud(background_color='white',colormap='inferno').generate(text)

fig, ax = plt.subplots(figsize=(7,3))
ax.imshow(wordcloud, interpolation='bilinear')
ax.axis('off')
```

```python
# making venn diagrams
A = set (["apple","banana","watermelon"])
B = set (["pumpkin",'bluberry','apple','key lime'])
```

```python
# Create the Venn diagram
diagram = venn2_unweighted(
    [A, B],
    set_labels=("Fruits", "Pies"),
    set_colors=("red", "blue"),
    alpha=0.5
)

# Get the labels for each subset of the diagram
labels = ['10', '01', '11']
subset_labels = [A - B, A & B, B - A]

# Add text to the subsets
for label, subset in zip(labels, subset_labels):
    # Get the coordinates of the center of the subset
    x, y = diagram.get_label_by_id(label).get_position()
    # Add the text using plt.text
    plt.text(x, y, '\n'.join(subset), ha='center', va='center')

plt.show()
```
